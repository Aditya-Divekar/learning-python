#  3.) Lambda funnction  / one line function / higher order function / no name function

```python
# Lambda functions (Anonymous functions): 
# Lambda functions are small, anonymous functions that are defined using the lambda keyword. 
# They can take any number of arguments but can only have a single expression in their body. 
# Lambda functions are commonly used when a small, one-time function is needed without the need for a formal function definition.
# Example: lambda x, y: x + y
```

```python
# What is a Lambda Function?
# A lambda function is a small, 
# one-line anonymous function
# It has no nameIt contains only one expression
# is a function that is defined without name
```

```python
#  syntax
# lambda arguments : expressioin
```

```python
add = lambda x,y : x+y
print(add(200,400))
```

```python
f = lambda a,b : a+b
result = f(10,20)
```

```python
print(result)
```

```python
a = lambda x,y : x if x>0 else y
a(20,30)
```

```python
#  MAP  : USED FOR MODIFICATION
# FILTER ... SELECT
# REDUCE ... COMBINES
```

```python
# FILTER 
numbers = [1,2,3,4,5,6,7,8]
result = list(filter(lambda x: x% 2== 0 , numbers))
print(result)
```

```python
# map
nums = [1,2,3,4,5,6,7,8]
result = list(map(lambda x: x*2, nums))
```

```python
print(result)
```

# REDUCE


```python
from functools import reduce
numbers = [1,2,3,4]
result = reduce(lambda a,b : a+b , numbers)
```

```python
print(result)
```

# GLOBAL AND LOCAL VARIABLE

```python
# local : which can access only in a function
# GLOBAL :  ACCESS EENY WHERE IN PROGRAMM
```

```python
num = 12 
def show():
    print("inside",num)
show()
print("outside",num)
```

```python
num = 12 
def show():
    num  = 34
    print("inside",num)
show()
print("outside",num)
```

```python
# How to make local variable global
num = 12 
def show():
    global num
    num  = 34
    print("inside",num)
show()
print("outside",num)
```

```python
# 4) Variable name argument ...args
# collects data into tuple
def add_numbers(*args):
    print(args)
add_numbers(10,20,30)
```

```python
# keyword Variable name argument ...args
# collects data into key val pair
def person(**kwargs):
    print(kwargs)
person(name = "adi", age = 18 , city = "pune")
```

```python
# 4. Function call itself called as recursion

def factorial(n):
    if n < 1 :
        return 1
    else :
        return n * factorial(n-1)

factorial(4)
```

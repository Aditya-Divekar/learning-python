# 1.Iterator

```python
# an iterator is an obj that allows u you to iterate elements one by one
# it provides on demand iteration
# using following methods

# __iter__()
# __next__()
```

```python
num = [10,20,30,40]
for i in num:
    print(i)
```

```python
num = [10,20,30,40]
# convert list into iterator

it = iter(num)
print(next(it))
```

```python
print(next(it))
print(next(it))
print(next(it))
```

# Generator

```python
# it is also an iterator 
# on demand iteration
# its a function that acts like a iterator
# it generates vaalues one by one autommatically

# Why Use Generator?

# ✔️ Less memory usage
# ✔️ Faster for large data
# ✔️ Easier than custom iterator
```

```python
def numbers() :
    yield 1
    yield 2
    yield 3
g = numbers()
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

# Swapping

```python
# exchanging values of two variables
a = 10
b = 20
temp = a
a = b
b = temp

print(a,b)
```

```python
#  swappign without using 3rd variable

a = 100
b = 200
a,b = b,a
print(a,b)
```

```python
# Getter Method

#  Definition:
# A getter method is used to access (get/read) the value of a variable from an object.
# It does not modify the data—only returns it.
# def get_age(self):
#     return self.age
# print(s1.get_age())

# Setter Method

# Definition:
# A setter method is used to modify (set/update) the value of a variable in an object.

# It allows controlled updating of data.
# def update_age(self, value):
 #   self.age = value
# s1.update_age(30)
```

```python
class Student:
    school = 'DPS'   # Class variable (shared by all objects)

    def __init__(self, name, age):
        self.name = name    # Instance variable
        self.age = age

    # Method to display details
    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, School: {Student.school}")

    # Getter method
    def get_age(self):
        return self.age

    # Setter method
    def update_age(self, value):
        self.age = value


# Object creation
s1 = Student('Avi', 23)
s2 = Student('Raj', 25)

# Access class variable
print(s1.school)        # via object
print(Student.school)   # via class

# Call methods
s1.show()
s2.show()

# Get age
print("Age of s1:", s1.get_age())

# Update age
s1.update_age(30)
print("Updated Age of s1:", s1.get_age())
```

# 1. INHERITANCE

```python
# the properties of one class is derived into another class
# 1 single level
# 2 multi level
# 3 multiple
```

```python
# single level # super class and sub class concept in python
# A.. parent class/super class/base class
# .
# .
# B..child class/sub class/derived class
```

```python
class Person:
    def _init_(self, name):
        self.name = name

    def show(self):
        print("Name:", self.name)


class Student(Person):   # inherits Person
    def __init__(self, name, marks):
        super()._init_(name)
        self.marks = marks

    def display(self):
        print("Marks:", self.marks)


s = Student("Avi", 90)
s.show()
s.display()
```

```python
# 2. Multi-Level Inheritance
# 👉 Grandparent → Parent → Child
# 👉 Vehicle → Car → SportsCar

class Vehicle:
    def start(self):
        print("Vehicle starts")


class Car(Vehicle):
    def drive(self):
        print("Car is driving")


class SportsCar(Car):
    def speed(self):
        print("Sports car is very fast")


c = SportsCar()
c.start()
c.drive()
c.speed()
```

```python
# 3. Multiple Inheritance
# 👉 Father + Mother → Child
# 👉 Teacher + Researcher → Professor

class Father:
    def skills(self):
        print("Father: Driving")


class Mother:
    def skills(self):
        print("Mother: Cooking")


class Child(Father, Mother):
    def own(self):
        print("Child: Coding")


c = Child()
c.skills()   # method resolution order (MRO)
c.own()
```

# 3. ENCAPSULATION

```python
# idea of rapping the data to avoid modification of data
# to put restriction f7on accessing variabale and method direcctly
```

```python
# public member
# Accessible everywhere
# private member
# Accessible inside class & child class
# protected member
# Accessible only inside class

class Student:
    name = "Avi"   # public variable

    def show(self):
        print("Name:", self.name)


s = Student()

print(s.name)
s.show()
```

```python
# private member
# only in class
# no outside class,no child class
# __ double unde

class BankAccount:
    __balance = 1000   # private variable

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount


# Object
acc = BankAccount()

# Access using getter
print(acc.get_balance())

# Update using setter
acc.deposit(500)

print(acc.get_balance())
```

```python
# protected member
# only in a class and its subclass
# not outside 
# _ single underscore

class Employee:
    _salary = 50000   # protected variable

    def show(self):
        print("Salary:", self._salary)


class Manager(Employee):
    def display(self):
        print("Manager Salary:", self._salary)


m = Manager()

m.show()
m.display()

print(m._salary)
```

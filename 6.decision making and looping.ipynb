```python
# takes ammount from user ammount >= 1000 ... 10% discount
# otherwise it gives 5% discount

amm = int(input("Enter the amount"))

amm10 = amm - amm * (10/100)
amm5 = amm - amm * (5/100)

if amm > 1000:
    print("10% discounted price ",amm10)
else :
    print("5 % discounted price ",amm5)
```

```python
# discount , total amm , payable amm from customer
# amm > 5000 ... 20 % discount
# amm >= 3000 and amm <= 5000 ...10% discount
# no discount

tot_amount = int(input("Enter total amount: "))

if tot_amount > 5000:
    discount = tot_amount * 0.20
elif tot_amount >= 3000 and tot_amount <= 5000:
    discount = tot_amount * 0.10
else:
    discount = 0

payable_amount = tot_amount - discount

print(f"Total Amount:{tot_amount}")
print(f"Discount:{discount}")
print(f"Payable Amount:{ payable_amount}")
```

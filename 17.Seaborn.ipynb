```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Aditya\OneDrive\Desktop\FORTUNE CLOUD\PYTHON\students.csv")
```

```python
df
```

# Linechart

```python
sns.lineplot(x= df['Study hrs'], y=df['Percantege'])
plt.title('Study hrs vs percentage')
plt.show()
```

```python
sns.lineplot(x= df['Study hrs'], y=df['Percantege'], data = df , hue="Class")
plt.title('Study hrs vs percentage')
plt.show()
```

```python
list1 = [11,22,33,44,50,51]
max1 = list1[0]
for i in list1:
    if i >= max1 :
        max1 = i
print(max1)
```

```python
list1 = [11,22,33,44,50,4,51]
min1 = list1[0]
for i in list1:
    if i < min1 :
        min1 = i
print(min1)
```

```python
# list1 = [11,22,33,44,50,51]

for i in range(len(list1)):
    for j in range(i+1, len(list1)):
        if list1[i] < list1[j]:
            list1[i], list1[j] = list1[j], list1[i]

print(list1)
```

```python

# List comprehension 

num = []
for i in range(5):
    num.append(i)
print(num)
```

```python
num = [i for i in range(5)]
print(num)
```

```python
even = []
for i in range(11):
    if i % 2 == 0:
        even.append(i)
print(even)
```

```python
even = [i for i in range(10) if i%2 == 0]
print(even)
```

# While and For loops

```python
# TYPES OF LOOOPS 
    1.WHILE LOOP = REPETADELY EXECUTES A BLOCK OF CODE AS GIVEN CONDITION IS TRUE
                   USED WHEN NO OF ITERATION IS UNKNOWN
    2.FOR LOOP =  IS USED TO ITERATE OVER A SEQUENCE SUCH AS LIST , TUPLE , STRING
                  USED WHEN NO. OF ITERATIONS ARE KNOWN
```

1. # WHILE LOOP

```python
SYNTAX : 
            INITIALIZATION
            WIHLE : CONDITION
                STATEMENT
                INCREMENT
```

```python
count = 1
while (count <= 5) :
    print("assdfghjkl")
    count += 1
```

```python
# PRINT FIRST 10 NO. 

num = 1 
while num<= 10:
    print(num, end = " ")
    num += 1
```

```python
# PRINT FIRST 6 EVEN NO. 
count = 1
num = 2
while count < 7:
    print(num , end = " ")
    num += 2
    count += 1
```

```python
count = 1
fact = 1 
while count <=5 :
    fact = fact*count
    count += 1
    print(fact)
```

```python
# loop control statements
# 1.break
    # it stops next all iterations it exits the loop now

# 2.pass
    # pass is a null statement in python it instructs to do nothing

# 3.continue
    # it skips particular iterations, in given program
```

```python
list1 = [10,20,30,40,50]
for i in list1:
    if i == 30 :
        pass
    print(i)
```

```python
for i in range(1,20):
    pass
```

# Break

```python
for i in range(2,9):
    if i > 6:
        break
    print(i)
```

#  Continue

```python
for i in range(2,9):
    if i == 6:
        continue
    print(i)
```

```python
name = "lmkj"
for i in name:
    if i == "m" :
        continue
    print(i)
```

```python
x = ["banana","apple","grapes","mango","banana"]
for i in x:
    if i == "mango" or i == "Mango":
        continue
    print(i)
```

```python
# prime no. program

num = int(input("enter the no"))
for i in range(2,num):
    if num % 2 == 0:
        print("not prime no")
        break
else :
    print(f"{num} is a prime no")
```

```python
# fibonacci number

num1 = 0
num2 = 1
print(num1 , end = " ")
print(num2 , end = " ")
for i in range(0,11):
    result = num1 + num2
    print(result , end = " ")
    num1 = num2
    num2 = result
```

```python
# zip() function
# Used to combine multiple sequence data types
# like lists, tuples, or strings into one iterable.
# It creates tuples by taking elements from the same position.
# The zipped result can be converted into list() or dict().
```

```python
# zip function
```

```python
x= [1,2,3]
y= [4,5,6]
zip(x,y)
```

```python
for i,j in zip(x,y):
    print(i,j)
```

```python
 x = [10,20,30]
 y = [40,50,60]
 z = [70,80,90]
zip(x,y,z)
```

```python
for i ,j,k in zip(x,y,z):
    print(i,j,k)
```

```python
name=["avi","raj","ram"]
marks = [90,80,70]
zip(name,marks)
```

```python
# enumerate() function
# Used to get index and value together while iterating.
# It returns tuples containing (index, value).
# enumerate() mainly takes two arguments:
# 1. iterable
# 2. start value (optional)
```

```python
for i in enumerate("suraj"):
    print(i,j)
```

```python
name=["avi","raj","ram"]
subject =["math","english","hindi"]
marks = [90,80,70]
for i, (name,subject,marks) in enumerate((name,subject,marks)):
    print(i,name,subject,marks)
```

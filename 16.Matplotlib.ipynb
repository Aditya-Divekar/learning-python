# Matplotlib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
days = [1,2,3,4,5,6]
tempreture = [23,54,32,24,23,45]
```

```python
plt.plot(days,tempreture , color = "red" , marker = "o" , linestyle = "dotted" , linewidth = 5 ,markersize = 19)
plt.xlabel("Days")
plt.ylabel("Temp")
plt.grid(linestyle = "dashed" , linewidth = 1 , color = "grey")
plt.title("DAYS VS TEMP")
plt.show()
```

```python
days = [1,2,3,4,5,6]
pune_temp = [23,28,32,24,23,15]
mumbai_temp = [20,34,24,54,19,33]
```

```python
plt.plot(days,pune_temp , color = "blue" , marker = "o" , label = "Pune")
plt.plot(days,mumbai_temp , color = "green" , marker = "o" , label = "mumbai")
plt.legend(loc = 2)
plt.grid( linestyle = 'dashed')
plt.show()
```

```python
data = {
    "month": ['jan','feb','mar','apr','may','june'],
    "sales": [200,300,340,503,600,530],
    "profit" : [20,30,60,40,70,50]
}
```

# 1.Line Chart

```python
df = pd.DataFrame(data)

plt.plot(df['month'], df['sales'] , color = "blue" , marker = 'o' , label = 'sales')
plt.title("Monthly Sales")
plt.xlabel("months")
plt.ylabel("sales")
plt.legend(loc= 1)
plt.grid(linestyle = "dashed")
plt.show()
```

# 2.Bar Chart

```python
plt.bar(df['month'] , df['sales'] ,  color = "red" , label = 'sales')
plt.title('Monhly sales')
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(linestyle = "dashed")
plt.legend(loc = 1)
plt.show
```

# 3.Horizontal Bar Chart

```python
plt.barh(df['month'],df['sales'],color = "yellow" , label = 'sales')
plt.title("Sales by Month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()
plt.legend(loc= 4)
plt.show()
```

# 4.Pie Chart

```python
plt.pie(df['sales'] , labels =df['month'] , autopct = "%1.1f%%")
plt.title('sales share')
plt.show()
```

# 5.Histogram

```python
plt.hist(df['sales'] ,bins = 1)
plt.title("Sales Distribution")
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```

# 6.Scatter Plot

```python
plt.scatter(df['sales'] , df["profit"] ,label= "datapoints")
plt.title("Sales vs profit")
plt.xlabel('Sales')
plt.ylabel('profit')
plt.grid()
plt.legend()
plt.show()
```

# 7.Boxplot

```python
plt.boxplot(df['sales'])
plt.title('Sales Spread and Outlier')
plt.ylabel('Sales')
plt.grid()
plt.show()
```

# Multiple Line Chart

```python
plt.plot(df['month'] , df['sales'] , marker = 'o' , label="sales")
plt.plot(df['month'] , df['profit'] , marker = 'o' , label="Profit")
plt.title("Month vs Sales vs Profit")
plt.xlabel("Months")
plt.ylabel("Values")
plt.grid()
plt.legend(loc=5)

plt.figure(figsize = (8,4))
plt.show()
```

```python
data = {
    "Team": ["MI","CSK","RCB","KKR","MI","CSK","RCB","KKR"],
    "Opponent": ["CSK","MI","KKR","RCB","RCB","KKR","MI","CSK"],
    "Runs": [180,175,200,190,210,160,195,170],
    "Wickets": [5,6,4,7,3,8,5,6],
    "Venue": ["Mumbai","Chennai","Bangalore","Kolkata",
              "Mumbai","Chennai","Bangalore","Kolkata"],
    "Result": ["Win","Loss","Win","Loss","Win","Loss","Win","Loss"]
}

df = pd.DataFrame(data)

# IPL Team Colors
team_colors = {
    "MI": "blue",
    "CSK": "yellow",
    "RCB": "red",
    "KKR": "purple"
}

# Create colors list
colors = [team_colors[team] for team in df["Team"]]

# Figure background
plt.figure(figsize=(10,6), facecolor='black')

# Plot area background
ax = plt.gca()
ax.set_facecolor('black')

# Bar Chart
plt.bar(df["Team"], df["Runs"], color=colors)

# Title and Labels
plt.title("IPL Team Runs", color='white', fontsize=16)

plt.xlabel("Teams", color='white')
plt.ylabel("Runs", color='white')

# Axis color
plt.xticks(color='white')
plt.yticks(color='white')

# Grid
plt.grid(axis='y', color='gray', linestyle='--')

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Mean and Standard Deviation
mu = 50 
sigma = 10

# Generate values
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)

# Normal distribution formula
y = (1/(sigma * np.sqrt(2*np.pi))) * np.exp(-0.5 * ((x - mu)/sigma)**2)

# Plot
plt.plot(x, y)
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Normal Distribution Curve")
plt.show()
```

```python
# Generate normal distribution data
data = np.random.normal(loc=50, scale=10, size=1000)

# Plot histogram
plt.hist(data, bins=30, density=True)
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("Normal Distribution Curve")

plt.show()
```

```python
data = np.random.normal(50, 10, 1000)

plt.figure(figsize=(8,5))
sns.histplot(data, kde=True)

plt.title("Symmetrical Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid()

plt.show()
```

```python
data = np.random.exponential(scale=2, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(data, kde=True)

plt.title("Right Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid()

plt.show()
```

```python
# Symmetrical
sym = np.random.normal(50, 10, 1000)

# Right Skewed
right = np.random.exponential(scale=2, size=1000)

# Left Skewed
left = -np.random.exponential(scale=2, size=1000)

plt.figure(figsize=(15,5))

# Symmetrical
plt.subplot(1,3,1)
sns.histplot(sym, kde=True)
plt.title("Symmetrical Distribution")
plt.grid()

# Right Skewed
plt.subplot(1,3,2)
sns.histplot(right, kde=True)
plt.title("Right Skewed Distribution")
plt.grid()

# Left Skewed
plt.subplot(1,3,3)
sns.histplot(left, kde=True)
plt.title("Left Skewed Distribution")
plt.grid()

plt.tight_layout()
plt.show()
```
